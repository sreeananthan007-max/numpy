```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("fitness_tracker.csv")
print(df)
```

```python
print("null values")
print(df.isnull().sum())
```

```python
df["Steps"]=df["Steps"].fillna(df["Steps"].mean()) 
df["Calories_Burned"]=df["Calories_Burned"].fillna(df["Calories_Burned"].mean()) 
df["Heart_Rate"]=df["Heart_Rate"].fillna(df["Heart_Rate"].mean()) 
df["Sleep_Hours"]=df["Sleep_Hours"].fillna(df["Sleep_Hours"].mean()) 
df["Water_Intake"]=df["Water_Intake"].fillna(df["Water_Intake"].mean())
df=df.drop_duplicates()
df.columns=df.columns.str.strip()
print(df)
print("null values")
print(df.isnull().sum())
```

```python
plt.figure()
plt.plot(df["Day"],df["Calories_Burned"],marker="o",color="red",label="Calories Burned")
plt.title("Calories Burned per day")
plt.xlabel("Day")
plt.ylabel("calories")
plt.legend()
plt.show()
plt.figure(figsize=(10,5))
plt.bar(df["Day"],df["Steps"],color="blue",label="Steps")
plt.title("Steps per day")
plt.xlabel("Day")
plt.ylabel("Steps")
plt.legend()
plt.show()
```

```python
print("day with highest calories burn:")
high=df.loc[df["Calories_Burned"].idxmax()]
print(high["Day"])
print("day with lowest calories burn:")
low=df.loc[df["Calories_Burned"].idxmin()]
print(low["Day"])
best=df[["Steps","Calories_Burned","Sleep_Hours","Water_Intake"]].mean()
print("best day:")
print(best.idxmax())
```
