```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df['Age']
```

```python
df["Salary"]= df["Salary"].fillna(df["Salary"].median())
df["Salary"]
```

```python
print("orginaldata")
print(df)
cleandata=df.drop_duplicates(inplace=True)
print("cleandata")
print(cleandata)
```

```python
df["Department"] = df["Department"].str.strip()
df["Department"]
```

```python
df["Gender"] = df["Gender"].str.upper()
df["Gender"]
```

```python
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
df["JoiningDate"].dtypes
```

```python
import numpy as np
df["Salary"] = np.where(df["Salary"] < 0, np.nan, df["Salary"])
df["Salary"]
```

```python
df.to_csv("employees_cleaned.csv", index=False)
```
