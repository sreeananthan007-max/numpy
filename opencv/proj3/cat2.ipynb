```python
import cv2
```

```python
img = cv2.imread("cat.jpeg")
if img is None:
    print("Image not found!")
    exit()
print("Image loaded successfully")
```

```python
cv2.imshow("Original Image",img)
cv2.waitKey(0)
```

```python
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv2.imshow("Grayscale",gray)
cv2.waitKey(0)
```

```python
blur=cv2.blur(gray,(5,5))
cv2.imshow("Normal Blur", blur)
cv2.waitKey(0)
```

```python
gaussian=cv2.GaussianBlur(gray,(5, 5),0)
cv2.imshow("Gaussian Blur",gaussian)
cv2.waitKey(0)
```

```python
median=cv2.medianBlur(gray,5)
cv2.imshow("Median Blur",median)
cv2.waitKey(0)
```

```python
_,threshold = cv2.threshold(gray,127,255,cv2.THRESH_BINARY)
cv2.imshow("Threshold", threshold)
cv2.waitKey(0)
```

```python
canny=cv2.Canny(gray,100, 200)
cv2.imshow("Canny Edge",canny)
cv2.waitKey(0)
```

```python
sobel_x=cv2.Sobel(gray,cv2.CV_64F,1,0,ksize=3)
sobel_y=cv2.Sobel(gray,cv2.CV_64F,0,1,ksize=3)
sobel = cv2.magnitude(sobel_x,sobel_y)
cv2.imshow("Sobel Edge",sobel)
cv2.waitKey(0)
```

```python
laplacian=cv2.Laplacian(gray,cv2.CV_64F)
cv2.imshow("Laplacian Edge", laplacian)
cv2.waitKey(0)
```

```python
inverted = cv2.bitwise_not(gray)
blurred = cv2.GaussianBlur(inverted,(21, 21),0)
sketch = cv2.divide(gray,255 - blurred,scale=256)
cv2.imshow("Pencil Sketch", sketch)
cv2.waitKey(0)
```

```python
cv2.imwrite("Cat_gray.png", gray)
cv2.imwrite("Cat_blur.png", blur)
cv2.imwrite("Cat_gaussian.png", gaussian)
cv2.imwrite("Cat_median.png", median)
cv2.imwrite("Cat_Cat_threshold.png", threshold)
cv2.imwrite("Cat_canny.png", canny)
cv2.imwrite("Cat_sobel.png", sobel)
cv2.imwrite("Cat_laplacian.png", laplacian)
cv2.imwrite("Cat_sketch.png", sketch)
print("All Cat images saved successfully!")
```
