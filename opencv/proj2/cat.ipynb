```python
import cv2
import numpy as np
```

```python
img = cv2.imread("cat.jpeg")
if img is None:
    print("Image not found!")
    exit()
print("Image loaded successfully")
```

```python
rotated=cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
cv2.imshow("Cat Rotated Image", rotated)
cv2.waitKey(0)
```

```python
resized = cv2.resize(img,(500, 300))
cv2.imshow("Cat Resized Image",resized)
cv2.waitKey(0)
```

```python
flipped = cv2.flip(img,1)
cv2.imshow("Flipped Image",flipped)
cv2.waitKey(0)
```

```python
cropped = img[100:400, 100:500]
cv2.imshow("Cat Cropped Image", cropped)
cv2.waitKey(0)
```

```python
height,width = img.shape[:2]
matrix = np.float32([[1, 0, 100],
                     [0, 1, 50]])
translated = cv2.warpAffine(img, matrix, (width,height))
cv2.imshow("Cat Translated Image", translated)
cv2.waitKey(0)
```

```python
cv2.imwrite("cat_rotated.png", rotated)
cv2.imwrite("cat_resized.png", resized)
cv2.imwrite("cat_flipped.png", flipped)
cv2.imwrite("cat_cropped.png", cropped)
cv2.imwrite("cat_translated.png", translated)
print("All images saved successfully")
```
