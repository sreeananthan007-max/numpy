```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
np.random.seed(10)
day=["day1","day2","day3","day4","day5","day6","day7","day8","day9","day10"]
dt=np.random.randint(25,41,10)
nt=np.random.randint(13,31,10)
df=pd.DataFrame({"days":day,"day temp":dt,"night temp":nt})
print(df)
plt.figure()
plt.plot(df.index,df["days"],df["day temp"],marker="o",label="day temp")
plt.plot(df.index,df["days"],df["night temp"],marker="o",label="night temp")
plt.title("day & night temp")
plt.xlabel("days")
plt.ylabel("temp")
plt.legend()
plt.show()
ad=df["day temp"].mean()
an=df["night temp"].mean()
plt.figure()
plt.bar(["Day Average", "Night Average"], [ad,an])
plt.title("Avg Temp")
plt.xlabel("Temp Type")
plt.ylabel("Temp")
plt.show()
hottest = df.loc[df["day temp"].idxmax()]
print("Hottest Day")
print(hottest)
coolest = df.loc[df["night temp"].idxmin()]
print("\nCoolest Day")
print(coolest)
```
