```python
import pandas as pd
```

```python
data={"Name": ["Alice", "Bob", "Alice", "David"],
        "Age": [25, 30, 25, 40],
        "City": ["NY", "LA", "NY", "Chicago"]}

df=pd.DataFrame(data)
print(df)
print('--------------------------------------------------------------------------------------------------------------------------')
df_cl=df.drop_duplicates()
print(df_cl)
```

```python
df=pd.DataFrame({'Name': ['Alice', 'Bob', 'Alice', 'David'],
                  'Age': [25, 30, 25, 40],
                  'City': ['NY', 'LA', 'SF', 'Chicago']})
df_cl=df.drop_duplicates(subset=["Name"])
print(df_cl)
```

```python
df = pd.DataFrame({   'Name': ['Alice', 'Bob', 'Alice', 'David'],
                'Age': [25, 30, 25, 40],
                'City': ['NY', 'LA', 'NY', 'Chicago']})
df_cl= df.drop_duplicates(keep='last')
print(df_cl)
```

```python
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Alice', 'David'],
                    'Age': [25, 30, 25, 40],
                    'City': ['NY', 'LA', 'NY', 'Chicago']})
df_cl=df.drop_duplicates(keep=False)
print(df_cl)
```

```python
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Alice', 'David'],
                 'Age': [25, 30, 25, 40],
                 'City': ['NY', 'LA', 'NY', 'Chicago']})
df.drop_duplicates(inplace=True)
print(df)
```

```python
import pandas as pd

data = {"Name": ["Alice", "Bob", "Alice", "David", "Bob"],
        "Age": [25, 20, 25, 40, 30],
        "City": ["NY", "LA", "NY", "Chicago", "LA"]}
df=pd.DataFrame(data)
df_cleaned=df.drop_duplicates(subset=["Name", "City"])
print(df)
print(df_cleaned)
```

```python
data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'], 
        'Age': [25, 30, 22, 35, 28], 
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'], 
        'Salary': [50000, 55000, 40000, 70000, 48000]}
df=pd.DataFrame(data)
print(df.dtypes)
df['Age']=df['Age'].astype(float)
print(df.dtypes)
```

```python
import numpy as np
df = pd.DataFrame({"FirstName": ['Vipul', 'Ashish', 'Milan'],
                    "Gender": ["", "", ""],
                    "Age": [0, 0, 0]})
df['Department'] = np.nan    
print(df)
```

```python
df = pd.DataFrame({"FirstName":['Vipul', 'Ashish', 'Milan'],
                    "Gender":["", "", ""],
                    "Age":[0, 0, 0]})
df['Department'] = np.nan
display(df)
df.dropna(how='all',axis=1,inplace=True)
display(df)
```

```python
df = pd.DataFrame({'FirstName': ['Vipul', 'Ashish', 'Milan'],
                    "Gender": ["", "", ""],
                    "Age": [0, 0, 0]})
df['Department'] = np.nan
display(df)
nan_value=float("NaN")
k = df.replace("", nan_value)
display(k)
k.dropna(how='all',axis=1,inplace=True)
display(k)
```

```python
df = pd.DataFrame({"FirstName": ['Vipul', 'Ashish', 'Milan'],
                            "Gender": ["", "", ""],
                            "Age": [0, 0, 0]})
df['Department'] = np.nan
display(df)
nan_value=float("NaN")
df.replace(0, nan_value, inplace=True)
df.dropna(how='all', axis=1, inplace=True)
display(df)
```

```python
data_frame = pd.DataFrame([['tom', 10],['nick', '15'],['juli', 14.5]],columns=['Name', 'Age'])
print(data_frame.info())
for column in data_frame.columns:
    print(column,':',pd.api.types.infer_dtype(data_frame[column]))
```
