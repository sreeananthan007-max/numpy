```python
import numpy as np
```

```python
import pandas as pd
```

```python
employees={"EmpID": [101,102,103,104,105,106,107,108],
            "Name": ["Alice","Bob","Charlie","David","Eva","Frank","Grace","Helen"],
            "Department": ["HR","IT","IT","Sales","HR","Finance","Sales","Finance"],
            "Experience": [2,5,3,7,4,6,8,5],
            "Salary": [35000,60000,45000,70000,50000,65000,80000,62000],
            "Rating": [4.2,4.8,4.0,3.9,4.5,4.7,4.9,4.4]}
df=pd.DataFrame(employees)
```

```python
print("\nEmployee Data")
print(df)
print("\nFirst 5 Employees")
print(df.head())
print("\nLast 5 Employees")
print(df.tail())
print("\nData Information")
print(df.info())
print("\nStatistical Summary")
print(df.describe())
```

```python
salary = df["Salary"]
experience = df["Experience"]
department = df["Department"]
print("\nSalary Series")
print(salary)
print("\nAverage Salary")
print(salary.mean())
print("\nMaximum Salary")
print(salary.max())
print("\nMinimum Salary")
print(salary.min())
print("\nAverage Experience")
print(experience.mean())
```

```python
salary_array=np.array(df["Salary"])
print("\nNumpy Salary Array")
print(salary_array)
print("\nSalary Mean")
print(np.mean(salary_array))
print("\nSalary Median")
print(np.median(salary_array))
print("\nSalary Standard Deviation")
print(np.std(salary_array))
print("\nSalary Variance")
print(np.var(salary_array))
```

```python
df["Bonus"] = df["Salary"] * 0.10
df["TotalSalary"] = df["Salary"] + df["Bonus"]
print("\nAfter Adding Bonus")
print(df)
```

```python
def salary_category(salary):
    if salary<40000:
        return "Low"
    elif salary<=60000:
        return "Medium"
    else:
        return "High"
df["SalaryCategory"]=df["Salary"].apply(salary_category)
print("\nSalary Category")
print(df)
```

```python
print("\nEmployees Salary>60000")
print(df[df["Salary"]>60000])
print("\nIT Department Employees")
print(df[df["Department"]=="IT"])
print("\nExperience Above 5 Years")
print(df[df["Experience"] > 5])
print("\nRating Above 4.5")
print(df[df["Rating"]>4.5])
```

```python
print("\nSorted By Salary")
print(df.sort_values("Salary", ascending=False))
print("\nSorted By Rating")
print(df.sort_values("Rating", ascending=False))
```

```python
print("\nAverage Salary Department Wise")
print(df.groupby("Department")["Salary"].mean())
print("\nHighest Salary Department Wise")
print(df.groupby("Department")["Salary"].max())
print("\nAverage Rating Department Wise")
print(df.groupby("Department")["Rating"].mean())
```

```python
df["Rank"] = df["Salary"].rank(ascending=False)
print("\nEmployee Ranking")
print(df[["Name","Salary","Rank"]])
```

```python
highest = df.loc[df["Salary"].idxmax()]
lowest = df.loc[df["Salary"].idxmin()]
print("\nHighest Paid Employee")
print(highest)
print("\nLowest Paid Employee")
print(lowest)
```

```python
print("\nTotal Company Salary")
print(df["Salary"].sum())
print("\nAverage Company Salary")
print(df["Salary"].mean())
```

```python
df["Salary"]=df["Salary"].astype(float)
df.loc[df["Department"]=="IT","Salary"]*= 1.15
print("\nAfter IT Salary Increment")
print(df)
```

```python
df = df[df["Rating"] >= 4]
print("\nAfter Removing Low Rating Employees")
print(df)
```

```python
name = input("\nEnter Employee Name: ")
result = df[df["Name"].str.lower() == name.lower()]
if len(result) > 0:
    print("\nEmployee Found")
    print(result)
else:
    print("\nEmployee Not Found")
```
