```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
np.random.seed()
elec=np.random.randint(5000,20001,10)
clot=np.random.randint(3000,15001,10)
groc=np.random.randint(2000,10001,10)
day=["day1","day2","day3","day4","day5","day6","day7","day8","day9","day10"]
df=pd.DataFrame({"days":day,"electronics":elec,"clothing":clot,"groceries":groc})
print(df)
plt.figure
plt.plot(df["days"],df["electronics"],marker="o",label="electronics")
plt.plot(df["days"],df["clothing"],marker="o",label="clothing")
plt.plot(df["days"],df["groceries"],marker="o",label="groceries")
plt.title("sales trend over 10 days")
plt.xlabel("days")
plt.ylabel("sales")
plt.legend()
plt.show()
catagories=["electronics","clothing","groceries"]
avg = df[["electronics", "clothing", "groceries"]].mean()
plt.figure
plt.bar(catagories,avg,color=["red","blue","green"])
plt.title("avg sales")
plt.xlabel("catagories")
plt.ylabel("avg")
plt.show()
df[total_sale]=df["electronics"]+df["clothing"]+df["groceries"]
print("best day of sale:")
print(df.loc[total_sale.idxmax(),"days"])
print("Lowest day:")
print(df.loc[total_sale.idxmin(),"days"])
print("top catagory:")
print(avg.idxmax())
```
