```python
x = np.array([1, 2, 3, 4])
y = x*2
plt.plot(x, y)
plt.xlabel("X-axis")        
plt.ylabel("Y-axis")        
plt.title("sree's graph")  
plt.show()
```

```python
x = np.array([1, 2, 3, 4])
y = x*2
plt.plot(x, y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("graph")
plt.show() 
x1 = [2, 4, 6, 8]
y1 = [3, 5, 7, 9]
plt.plot(x1, y1, '-.')
plt.show()
```

```python
x = np.array([1, 2, 3, 4])
y = x * 2
x1 = [2, 4, 6, 8]
y1 = [3, 5, 7, 9]
plt.plot(x, y, label='Line 1')
plt.plot(x1, y1, '-.', label='Line 2')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Both Lines in One Chart")
plt.legend()
plt.show()
```

```python
x = np.array([1, 2, 3, 4])
y = x*2
plt.plot(x, y)

x1 = [2, 4, 6, 8]
y1 = [3, 5, 7, 9]

plt.plot(x, y1, '-.')
plt.xlabel("X-axis data")
plt.ylabel("Y-axis data")
plt.title('multiple plots')

plt.fill_between(x, y, y1, color='green', alpha=.5)
plt.show()
```
