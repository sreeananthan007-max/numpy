```python
import pandas as pd
data=pd.DataFrame({"genre": ["Pop","Pop","Rock","Rock","Jazz","Jazz"],
                   "gender": ["Male","Female","Male","Female","Male","Female"],
                   "count": [50,60,40,55,30,45]})
sns.barplot(x="genre",y="count",data=data,hue="gender")
sns.set(style="darkgrid")
plt.title("Music Genre Popularity by Gender")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
data=sns.load_dataset("tips")
sns.histplot(data=data,x="total_bill",bins=10,kde=True,color="skyblue")
plt.title("Histogram of Total Bill (Seaborn)")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
data = {"Name":["ANSH","SAHIL","ANSH","JAYAN","ANURAG","ANURAG","ANURAG","SAHIL"]}
df = pd.DataFrame(data)
sns.countplot(x="Name",data=df)
plt.title("Frequency of Names")
plt.show()
```

```python
from sklearn.datasets import load_iris
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
iris=load_iris()
df=pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"]=iris.target
df["Species"]=df["Species"].map({ 0:"Setosa",1:"Versicolor",2:"Virginica"})
sns.kdeplot(data=df[df["Species"]=="Virginica"],x="sepal length (cm)",fill=True,label="Virginica")
plt.legend()
plt.show()
```
