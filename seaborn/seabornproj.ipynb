```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed()
order_id=np.arange(1, 201)
day = np.random.choice(["Day1","Day2","Day3","Day4","Day5","Day6","Day 7",
                        "Day8","Day9","Day10","Day11","Day12","Day13","Day 14"],200)
category=np.random.choice(["Electronics", "Clothing", "Groceries", "Books"],200)
payment=np.random.choice(["UPI", "Card", "COD", "Wallet"],200)
status=np.random.choice(["Delivered","Cancelled","Returned","Processing"],200)
order_value=np.random.randint(500, 8001, 200)
df=pd.DataFrame({"order_iD":order_id,"Day":day,"product category":category,"payment method":payment,"order status":status,"order value": order_value})
print(df)
sns.countplot(x="payment method", hue="order status", data=df)
plt.title("Order Status by Payment Method")
plt.xlabel("payment method")
plt.ylabel("number of orders")
plt.show()
avg=df.groupby("product category")["order value"].mean()
print(avg)
plt.bar(avg.index, avg.values)
plt.title("Average Order Value by Product Category")
plt.xlabel("product category")
plt.ylabel("average order value")
plt.show()
payment_count=df["payment method"].value_counts()
print("Most Used Payment Method:")
print(payment_count.idxmax())
daily_sales=df.groupby("Day")["order value"].sum()
print(daily_sales)
best_day=daily_sales.idxmax()
print("Best Performing Day:")
print(best_day)
cancelled=df[df["order status"]=="Cancelled"]
cancel_count=cancelled["product category"].value_counts()
print(cancel_count)
print("Category with Most Cancellations:")
print(cancel_count.idxmax())
```
